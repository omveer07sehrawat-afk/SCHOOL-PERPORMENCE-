```python
#import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.plot(data["STUDENTS NAME"], data["PERCENTAGE"],marker = 'p')
plt.grid(1)
plt.xticks(rotation=60,color = 'green')
plt.ylabel("Percentage",color = 'green')
plt.title("Student Percentage")
plt.show()
```

```python
top10 = data.sort_values(by="PERCENTAGE").head(15)

plt.figure(figsize=(10,5))
plt.bar(top10["STUDENTS NAME"], top10["PERCENTAGE"],color = "red")

plt.xticks(rotation=60,color = 'blue')
plt.title("Top 15 Students")
plt.ylabel("Percentage")
plt.show()
```

```python
subjects = ["ENGLISH","HINDI","BUSINESS","ACCOUNT","ECONOMICS"]

avg_marks = data[subjects].mean()

plt.bar(avg_marks.index, avg_marks.values, color = "black")

plt.title("Average Marks by Subject")
plt.ylabel("Average Marks")
plt.show()
```

```python
explode = [0.1, 0, 0, 0, 0]
plt.pie(avg_marks.values,
        labels=avg_marks.index,
        autopct='%1.1f%%')
#        wedgeprops={'edgecolor':'black'})
plt.title("Subject Contribution")
plt.show()
```

```python
plt.hist(data["PERCENTAGE"], bins=10,color = 'brown')
plt.grid(2,color = 'blue')
plt.xlabel("Percentage",color = 'red')
plt.ylabel("Number of Students",color = 'red')
plt.title("Distribution of Percentage",color = 'blue')
plt.show()
```

```python
plt.scatter(data.index,
            data["ENGLISH"],
            color='red',
            label='English')
plt.scatter(data.index,
            data["HINDI"],
            color='green',
            label='Hindi')

plt.xlabel("Student Index")
plt.ylabel("Marks")
plt.title("English vs Hindi Marks")
plt.grid(1)
plt.legend()
plt.show()
```

```python
plt.plot(data["STUDENTS NAME"], data["TOTAL"], marker='o',color = 'red')

plt.xticks(rotation=75)
plt.ylabel("Total Marks")
plt.title("Student Total Marks")
plt.show()
```

```python
above90 = data[data["PERCENTAGE"] > 90]

plt.bar(above90["STUDENTS NAME"],
        above90["PERCENTAGE"],color = 'green')

plt.title("Students Above 90%")
plt.xticks(rotation=45)
plt.show()
```

```python
low = data.sort_values(by="PERCENTAGE").head(5)

plt.bar(low["STUDENTS NAME"],
        low["PERCENTAGE"],color = 'red')

plt.title("Lowest 5 Students")
plt.show()
```

```python
student = data[data["STUDENTS NAME"] == "GAYATRI"]

marks = student[subjects].iloc[0]

plt.bar(subjects, marks,color = 'pink')

plt.title("Gayatri Subject Marks",color = 'red')
plt.ylabel("Marks",color = 'blue')
plt.show()
```

```python
subjects = ["ENGLISH","HINDI","BUSINESS","ACCOUNT","ECONOMICS"]

data.set_index("STUDENTS NAME")[subjects].plot(
    kind='bar',
    stacked=True,
    figsize=(12,6)
)

plt.title("Subject-wise Marks")
plt.ylabel("Marks")
plt.show()
```

```python
topper = data.sort_values(by="PERCENTAGE",
                        ascending=False).head(1)

plt.bar(topper["STUDENTS NAME"],
        topper["PERCENTAGE"],color = 'yellow')

plt.title("Class Topper")
plt.show()
```

```python
data["MATHS"] = data["MATHS"].fillna(0)
```

```python
subjects = ["ENGLISH","HINDI","BUSINESS","ACCOUNT","ECONOMICS"]
avg_marks = data[subjects].mean()
explode = [0] * len(avg_marks)
highest_index = avg_marks.argmax()
explode[highest_index] = 0.2
plt.figure(figsize=(8,8))

plt.pie(avg_marks.values,
        labels=avg_marks.index,
        autopct='%1.1f%%',
        explode=explode,
        shadow=True)

plt.title("Highest Average Subject Highlighted")

plt.show()
```

```python
subjects = ["ENGLISH","HINDI","BUSINESS","ACCOUNT","ECONOMICS"]
avg_marks = data[subjects].mean()
plt.figure(figsize=(8,8))
plt.pie(avg_marks.values,
        labels=avg_marks.index,
        autopct='%1.1f%%',
        startangle=90,
        wedgeprops={'edgecolor':'black'})
centre_circle = plt.Circle((0,0), 0.70, fc='white')
fig = plt.gcf()
fig.gca().add_artist(centre_circle)
plt.title("Subject Contribution Donut Chart")

plt.show()
```
